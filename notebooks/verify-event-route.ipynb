```python
import requests
path = "/api/events/"
base_url = "http://localhost:8002"
endpoint = f"{base_url}{path}"
print(endpoint)
```

```python
response = requests.get(endpoint)
if response.ok:
    data = response.json()
    print(data, type(data), data.get('items'))
```

```python
data = {"page": "/test+"}
```

```python
import json
json.dumps(data)
```

```python
headers = {
    "Content-Type": "application/json"
}
response = requests.post(endpoint, data=json.dumps(data), headers=headers)
print(response.headers)
print('ok', response.ok)
if response.ok:
    data = response.json()
    print(data, type(data), data.get('items'))
else:
    print(response.text)
```

```python
detail_path = "/api/events/12"
base_url = "http://localhost:8002"
detail_endpoint = f"{base_url}{detail_path}" 
print(detail_endpoint)

headers = {
    "Content-Type": "application/json"
}

response = requests.put(detail_endpoint, json={"description": "hello world"}, headers=headers)

print(response.headers)
print('ok', response.ok)
if response.ok:
    data_put = response.json()
    print(data_put, type(data_put))
else:
    print(response.text)
```
